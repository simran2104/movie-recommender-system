```python
# title
# release_date
# imdb_id
# overview
# genres
# spoken_languages
# cast
# director
# keywords
# popularity

movies = movies[['title', 'release_date', 'imdb_id', 'overview', 'spoken_languages', 'cast', 'director', 'keywords', 'popularity']]
```

```python
movies.head(1)
```

```python
movies.isnull().sum()
```

```python
movies = movies.dropna(subset=['imdb_id'])
```

```python
movies.duplicated().sum()
```

```python
movies['spoken_languages'] = movies['spoken_languages'].fillna('Unknown')
```

```python
movies['keywords'] = movies['keywords'].fillna('Unknown')
movies['cast'] = movies['cast'].fillna('Unknown')
movies['director'] = movies['director'].fillna('Unknown')
movies['overview'] = movies['overview'].fillna('Unknown')
```

```python
movies = movies.dropna(subset=['title'])
```

```python
movies['release_date'] = pd.to_datetime(movies['release_date'] , errors = 'coerce')
```

```python
movies.info()
```

```python
movies.head(1)
```

```python
movies['overview'] = movies['overview'].apply(lambda x: x.split())
```

```python
movies['spoken_languages'] = movies['spoken_languages'].apply(lambda x: x.split(','))
movies['cast'] = movies['cast'].apply(lambda x: x.split(','))
movies['keywords'] = movies['keywords'].apply(lambda x: x.split('|'))
```

```python
movies.head(1)
```

```python
movies['release_year'] = movies['release_date'].dt.year.astype('Int64')
movies = movies[['title', 'release_year', 'imdb_id', 'overview', 'spoken_languages', 'cast', 'director', 'keywords']]
```

```python
movies['cast'] = movies['cast'].apply(lambda x: [i.replace(" ", "") for i in x])
```

```python
movies['director'].str.replace(" ", "")
```

```python
movies['director'] = movies['director'].apply(lambda x: x.split(','))
```

```python
movies.head(1)
```

```python
movies['director'] = movies['director'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x: [i.replace(" ", "") for i in x])
```

```python
movies['spoken_languages'] = movies['spoken_languages'].apply(lambda x: [i.replace(" ", "") for i in x])
```

```python
movies['tags'] = movies['overview'] + movies['spoken_languages'] + movies['cast'] + movies['director'] + movies['keywords']
```

```python
movies_clean = movies[['title', 'imdb_id', 'tags']]
```

```python
movies_clean.head(1)
```

```python
movies_clean['tags'] = movies_clean['tags'].apply(lambda x:" ".join(x))
```

```python
movies_clean['tags'][0]
```

```python
movies_clean['tags'] = movies_clean['tags'].apply(lambda x: x.lower())
```

```python
movies_clean.head()
```

```python
movies_clean['tags'] = movies_clean['tags'].str.replace('Unknown','')
```

```python
# Now, we will do text vectorization
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features = 10000, stop_words='english', min_df=5)
```

```python
vectors = cv.fit_transform(movies_clean['tags'])
```
